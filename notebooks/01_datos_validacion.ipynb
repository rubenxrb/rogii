## 3. Limpieza de GR

**Decisiones de limpieza de GR**

- Cada fila equivale a 1 ft (`MD` avanza de 1 en 1 en los 773 pozos), así que huecos en filas = huecos en pies.
- Huecos de 1 a 9 ft: se interpolan linealmente (`GR_clean`). Son el 99.2 % de los huecos y el 93.6 % de las filas sin GR.
- Huecos de 10 ft o más y huecos en los bordes del pozo: no se inventan datos; quedan en NaN en `GR_clean` (96 310 de 1 507 972 filas faltantes). De los 70 huecos de 50 ft o más, 52 están en un borde del pozo.
- `GR_mask = True` donde el GR original faltaba, se haya interpolado o no. Antes del PS falta en promedio el 24 % del GR original (mediana 20 %) y desde el PS el 31 % (mediana 30 %): en promedio, el 31 % del tramo evaluado no tiene GR medido (los huecos cortos se interpolan; los largos y los de los bordes quedan en NaN).
- `clean_gr` nunca modifica un valor medido (verificado en los 773 pozos). El límite de 9 ft es el parámetro `max_gap`.
- Escala: el GR del horizontal y el del typewell están en la misma escala (media de las medias 88.1 y 88.7; ningún valor negativo). No se normaliza: `GR_clean` conserva las unidades originales y cada parte puede normalizar como necesite.
- Ruido: la razón std(saltos) / std total va de 0.05 a 0.60 (mediana 0.29) en los 773 pozos, o sea una señal continua. No se suaviza `GR_clean`; `smooth_gr` (mediana móvil centrada, ventana 9) queda como función opcional.
- Valores altos: no se recortan. De 3 pozos revisados con GR > 300, uno es un pico aislado (`a4f989c2`: 487 con vecinos en 118) y los otros dos son zonas de GR alto (vecinos en 315 y 200). `smooth_gr` atenúa los picos aislados.

```python
def clean_gr(hz, max_gap=9):
    """Interpola huecos cortos de GR y marca todo lo que faltaba.

    GR_clean: GR con los huecos de hasta max_gap filas interpolados (lineal).
              Los huecos más largos, y los que están en los bordes del pozo, siguen en NaN.
    GR_mask : True donde el GR original faltaba (se haya interpolado o no).
    """
    gr = hz["GR"]
    mask = gr.isna()

    # longitud del hueco al que pertenece cada fila vacía
    grupo = (~mask).cumsum()
    largo_hueco = mask.groupby(grupo).transform("sum")

    # interpola solo entre valores válidos (no extrapola en los bordes)
    interp = gr.interpolate(method="linear", limit_area="inside")

    # deja en NaN lo que cae en huecos largos
    gr_clean = interp.where(~(mask & (largo_hueco > max_gap)))

    out = hz.copy()
    out["GR_clean"] = gr_clean
    out["GR_mask"] = mask
    return out
```

```python
def smooth_gr(gr_clean, window=9):
    """Mediana móvil centrada. No rellena los huecos que quedaron en NaN."""
    suave = gr_clean.rolling(window, center=True, min_periods=1).median()
    return suave.where(gr_clean.notna())
```

## 4. Remuestreo del typewell

**Decisiones de remuestreo del typewell**

- Los 773 typewells tienen el TVT estrictamente creciente y no traen NaN en GR. El paso mediano es 0.5 ft en 653 pozos, 0.25 ft en 91, 1.0 ft en 21 y 0.10 ft en 8 (este último no figuraba en el notebook de referencia).
- La mayor parte de las diferencias de paso dentro de un typewell es redondeo de los decimales. Solo 3 typewells tienen un salto real de TVT (hasta 9 ft).
- Grilla común de 0.5 ft (`step`, parámetro de `resample_typewell`) con origen en el primer TVT de cada typewell. El GR se interpola linealmente; los saltos de más de 9 ft quedarían en NaN (ninguno lo supera). `Geology` no se interpola: cada punto toma la etiqueta del punto original anterior. No se pierde ninguna de las 43 etiquetas, y en promedio el 30.6 % de las filas de un typewell no tiene etiqueta.
- Costo de remuestrear: el error medio de GR al volver a los TVT originales equivale al 0.4 % del cambio típico entre puntos vecinos en los de 0.5 ft, 8 % en los de 0.10 ft, 11 % en los de 0.25 ft y 19 % en los de 1 ft.
- Cobertura: 13 pozos tienen un horizontal cuyo TVT sale del rango de su typewell, pero ninguno lo hace desde el PS (el tramo evaluado siempre queda cubierto en train). 10 de ellos comparten el mismo typewell (rango 11782.6 a 12285.1) y su horizontal baja entre 343 y 645 ft por debajo de su inicio. No se extrapola ni se inventan datos; la tabla maestra llevará las banderas `tw_cubre_pozo` y `tw_cubre_ps`.
- Hay 752 typewells distintos para 773 pozos; la tabla maestra guarda el typewell asignado.
- Los picos agudos del typewell pueden perder amplitud al remuestrear. La pérdida máxima es de 19.9 unidades de GR en los de 0.5 ft, 31.9 en los de 0.25 ft, 13.0 en los de 1 ft y 9.2 en los de 0.10 ft. En el caso de 0.5 ft revisado (`e730c32b`, GR de 110 a 224 en 3 ft), solo 25 de 2188 puntos tienen error mayor a 5. En los typewells de 0.5 ft, el desfase máximo respecto a su grilla nominal es de 0.25 ft, pequeño frente a los 15.9 ft de RMSE de referencia.
- El typewell de test solo trae TVT y GR (sin Geology); resample_typewell lo soporta y devuelve Geology vacío.

```python
def resample_typewell(tw, step=0.5, max_gap=9.0):
    """Typewell a grilla regular de TVT (paso `step`, origen en su primer punto).

    GR: interpolación lineal; los huecos de más de max_gap pies quedan en NaN.
    Geology: etiqueta del punto original anterior (solo si el typewell la trae).
    """
    t = tw["TVT"].to_numpy()
    grid = np.arange(t[0], t[-1] + 1e-9, step)

    j = np.clip(np.searchsorted(t, grid + 1e-6, side="right") - 1, 0, len(t) - 1)
    k = np.clip(j, 0, len(t) - 2)
    gap = t[k + 1] - t[k]

    gr = np.interp(grid, t, tw["GR"].to_numpy())
    gr[gap > max_gap + 1e-6] = np.nan
    if "Geology" in tw.columns:
        geo = tw["Geology"].to_numpy()[j]
    else:
        geo = np.full(len(grid), None, dtype=object)
    return pd.DataFrame({"TVT": grid, "GR": gr, "Geology": geo})
```

## 5. Tabla maestra de pozos

**Tabla maestra de pozos (`wells_df`)**

- Una fila por pozo de train (773), con 26 columnas más `az_sin` y `az_cos`: geometría, PS, typewell asignado y banderas de cobertura. Sin valores vacíos.
- Todo se calcula con las columnas de test (`MD`, `X`, `Y`, `Z`, `GR`, `TVT_input`, y `TVT`/`GR` del typewell), salvo `form_dom` y `form_frac` (necesitan `Geology`, que solo existe en train; son descriptivas) y `tw_cubre_pozo` y `tw_cubre_ps` (necesitan el TVT real). Verificado en los 3 pozos de test: el resto de las columnas coincide con train.
- `azimuth` = atan2(ΔX, ΔY) entre la primera y la última fila, en grados (0° = hacia +Y). Es circular y la mayoría de los pozos sigue dos ejes (NW–SE: 482 pozos; N–S: 281; solo 10 fuera de ellos): usar `az_sin` y `az_cos`. `desplaz_horiz` es la distancia en línea recta cabeza-fin, no el largo del pozo.
- `tw_id` identifica el typewell por su `TVT` y `GR`: hay 752 distintos para 773 pozos, y 10 pozos comparten el mismo.
- `form_dom` es la etiqueta de `Geology` más frecuente entre las filas con TVT conocido (antes del PS): `EGFDL` en 638 de 773 pozos (82.5 %). `form_frac` es la fracción de esas filas que cae en un tramo con etiqueta.
- `tw_cubre_pre_ps` indica si el typewell cubre el TVT conocido. En train coincide con `tw_cubre_pozo` (13 pozos no cubiertos) y ninguno falla desde el PS; en test no se puede verificar el tramo posterior al PS.
- El `fold` se agrega en la sección 6.

```python
import hashlib

def well_row(w, split="train"):
    hz, tw, ps = load_well(w, split)
    k = max(ps - 1, 0)
    t = tw["TVT"].to_numpy()
    conocido = hz["TVT_input"].iloc[:ps].to_numpy()
    dx = hz["X"].iloc[-1] - hz["X"].iloc[0]
    dy = hz["Y"].iloc[-1] - hz["Y"].iloc[0]

    # formación dominante: solo si el typewell trae Geology (train)
    form_dom, form_frac = None, np.nan
    if "Geology" in tw.columns:
        dentro = (conocido >= t[0]) & (conocido <= t[-1])
        idx = np.clip(np.searchsorted(t, conocido[dentro], side="right") - 1, 0, len(t) - 1)
        etiquetas = pd.Series(tw["Geology"].to_numpy()[idx]).dropna()
        dom = etiquetas.mode()
        form_dom = dom.iloc[0] if len(dom) else None
        form_frac = float(len(etiquetas) / max(len(conocido), 1))

    fila = {
        "well": w, "n": len(hz), "ps": ps, "n_pred": len(hz) - ps,
        "x0": hz["X"].iloc[0], "y0": hz["Y"].iloc[0], "z0": hz["Z"].iloc[0],
        "x_ps": hz["X"].iloc[k], "y_ps": hz["Y"].iloc[k],
        "x_end": hz["X"].iloc[-1], "y_end": hz["Y"].iloc[-1],
        "azimuth": float(np.degrees(np.arctan2(dx, dy)) % 360),
        "desplaz_horiz": float(np.hypot(dx, dy)),
        "md0": hz["MD"].iloc[0], "md_ps": hz["MD"].iloc[k],
        "tvt_ps": hz["TVT_input"].iloc[k],
        "tw_id": hashlib.md5(pd.util.hash_pandas_object(tw[["TVT", "GR"]], index=False).to_numpy().tobytes()).hexdigest()[:8],
        "tw_n": len(tw), "tw_step": round(float(np.median(np.diff(t))), 2),
        "tw_min": t[0], "tw_max": t[-1],
        "tw_cubre_pre_ps": bool(conocido.min() >= t[0] and conocido.max() <= t[-1]),
        "form_dom": form_dom, "form_frac": form_frac,
    }
    if "TVT" in hz.columns:
        fila["tw_cubre_pozo"] = bool(hz["TVT"].min() >= t[0] and hz["TVT"].max() <= t[-1])
        fila["tw_cubre_ps"] = bool(hz["TVT"].iloc[ps:].min() >= t[0] and hz["TVT"].iloc[ps:].max() <= t[-1])
    return fila

wells_df = pd.DataFrame([well_row(w) for w in wells])
rad = np.radians(wells_df["azimuth"])
wells_df["az_sin"], wells_df["az_cos"] = np.sin(rad), np.cos(rad)

print(wells_df.shape)
print(wells_df[["n", "ps", "n_pred", "desplaz_horiz", "azimuth", "tvt_ps", "form_frac"]]
      .describe().round(1).T[["mean", "min", "50%", "max"]])
print(wells_df["form_dom"].value_counts(dropna=False).head(8))
print("typewells distintos:", wells_df["tw_id"].nunique())
print("typewell no cubre | lo conocido:", (~wells_df["tw_cubre_pre_ps"]).sum(),
      "| todo el pozo:", (~wells_df["tw_cubre_pozo"]).sum(),
      "| desde el PS:", (~wells_df["tw_cubre_ps"]).sum())
```

## 6. Esquema de validación (folds espaciales)

**Esquema de validación**

- La distancia entre dos pozos es la mínima en planta (X, Y) entre sus trayectorias completas, no solo entre cabezas.
- Se toma un punto cada 25 ft, incluido el final de cada pozo. Como el muestreo puede sobrestimar la distancia hasta en 25 ft, se agrupan los pares a 525 ft o menos. Así no se omite ningún par a menos de 500 ft (puede incluir algunos un poco más lejanos).
- Los pozos vecinos, y los vecinos de sus vecinos, forman un **grupo** que va completo a un mismo fold.
- Los grupos se reparten del más grande al más pequeño en el fold con menos filas evaluadas (`n_pred`). El resultado es reproducible (`seed=42`).
- Las partes 2 y 3 validan el fold `k` con `fold == k` y entrenan con `fold != k`.
- Pozos que comparten typewell pueden quedar en folds distintos: el criterio de agrupación es solo espacial.

```python
from scipy.spatial import cKDTree
from scipy.sparse import coo_matrix
from scipy.sparse.csgraph import connected_components

paso = 25          # una muestra cada 25 ft de trayectoria
RADIO = 525.0      # 500 ft + margen por el muestreo

xy_full = []       # trayectorias completas (se usan para verificar)
xs, ys, ids = [], [], []
for i, w in enumerate(wells):
    tray = load_well(w)[0][["X", "Y"]].to_numpy()
    xy_full.append(tray)
    pts = np.vstack([tray[::paso], tray[-1:]])      # incluye el final del pozo
    xs.append(pts[:, 0]); ys.append(pts[:, 1])
    ids.append(np.full(len(pts), i))
xy = np.column_stack([np.concatenate(xs), np.concatenate(ys)])
ids = np.concatenate(ids)

pares = cKDTree(xy).query_pairs(RADIO, output_type="ndarray")
a, b = ids[pares[:, 0]], ids[pares[:, 1]]
cruz = a != b
d = np.hypot(*(xy[pares[cruz, 0]] - xy[pares[cruz, 1]]).T)
vec = (pd.DataFrame({"a": np.minimum(a[cruz], b[cruz]), "b": np.maximum(a[cruz], b[cruz]), "d": d})
         .groupby(["a", "b"], as_index=False)["d"].min())
print("pares de pozos vecinos:", len(vec))

def neighbor_groups(vec, n_wells, radio):
    """Etiqueta de grupo de cada pozo, uniendo los pares a distancia <= radio."""
    v = vec[vec["d"] <= radio]
    g = coo_matrix((np.ones(len(v)), (v["a"].to_numpy(int), v["b"].to_numpy(int))),
                   shape=(n_wells, n_wells))
    return connected_components(g, directed=False)[1]
```

```python
def make_folds(wells_df, grupo, k=5, seed=42):
    """Asigna cada grupo completo a un fold, equilibrando las filas evaluadas.

    grupo: etiqueta de grupo de cada fila de wells_df (mismo orden).
    Devuelve un DataFrame con columnas well, fold.
    """
    peso = wells_df["n_pred"].groupby(grupo).sum()
    # el orden aleatorio solo decide los empates
    orden = (peso.sample(frac=1, random_state=seed)
                 .sort_values(ascending=False, kind="stable"))
    carga = np.zeros(k)
    asignado = {}
    for g, p in orden.items():
        f = int(carga.argmin())
        asignado[g] = f
        carga[f] += p
    return pd.DataFrame({"well": wells_df["well"].to_numpy(),
                         "fold": [asignado[g] for g in grupo]})

K = 5
assert list(wells_df["well"]) == list(wells)      # mismo orden que los índices de vec
grupo = neighbor_groups(vec, len(wells), RADIO)

mayor = wells_df["n_pred"].groupby(grupo).sum().max() / wells_df["n_pred"].sum()
print(f"grupos: {grupo.max() + 1} | el mayor tiene {mayor:.1%} de las filas evaluadas")
if mayor > 1 / K:
    print("AVISO: el grupo mayor supera un fold ideal; los folds quedarán desbalanceados")

folds = make_folds(wells_df, grupo, k=K)
wells_df = wells_df.drop(columns="fold", errors="ignore").merge(folds, on="well", validate="1:1")
assert wells_df["fold"].nunique() == K, "no hay suficientes grupos para formar K folds"

print(wells_df.groupby("fold").agg(pozos=("well", "size"), filas_eval=("n_pred", "sum")))
```

## 7. Función de evaluación y validador del formato de predicción

**Evaluación y formato de predicción**

- RMSE global: sobre **todas las filas** desde el PS juntas, así que los pozos con más filas pesan más. No es lo mismo que el promedio de los RMSE por pozo.
- `validate_prediction` exige las columnas `well`, `idx`, `tvt_pred`, sin nulos, infinitos ni duplicados, con todas las filas desde el PS de cada pozo y sin sobrantes. Con `expected_wells` también detecta pozos completos que falten (por ejemplo, los de un fold).
- `rmse_by_well` valida antes de medir, para que una predicción incompleta no obtenga un RMSE engañoso.

```python
def validate_prediction(pred, wells_df, expected_wells=None):
    """Lanza ValueError si `pred` no cumple el formato. Devuelve True si cumple."""
    cols = ["well", "idx", "tvt_pred"]
    if sorted(pred.columns) != sorted(cols):
        raise ValueError(f"columnas {list(pred.columns)}; se esperaban {cols}")
    if len(pred) == 0:
        raise ValueError("la predicción está vacía")
    if pred[cols].isna().any().any():
        raise ValueError("hay valores nulos")
    if not all(isinstance(w, str) for w in pred["well"].unique()):
        raise ValueError("`well` debe ser texto")
    if not pd.api.types.is_integer_dtype(pred["idx"]):
        raise ValueError("`idx` debe ser entero")
    if not pd.api.types.is_numeric_dtype(pred["tvt_pred"]):
        raise ValueError("`tvt_pred` debe ser numérico")
    if not np.isfinite(pred["tvt_pred"].to_numpy(dtype=float)).all():
        raise ValueError("`tvt_pred` tiene valores infinitos")
    if pred.duplicated(["well", "idx"]).any():
        raise ValueError("hay filas (well, idx) duplicadas")

    info = wells_df.set_index("well")
    presentes = set(pred["well"])
    if presentes - set(info.index):
        raise ValueError(f"pozos desconocidos: {sorted(presentes - set(info.index))[:5]}")
    if expected_wells is not None and presentes != set(expected_wells):
        raise ValueError(f"faltan {len(set(expected_wells) - presentes)} pozos "
                         f"y sobran {len(presentes - set(expected_wells))}")

    for w, g in pred.groupby("well")["idx"]:
        ps, n = int(info.at[w, "ps"]), int(info.at[w, "n"])
        # sin duplicados: mínimo, máximo y conteo garantizan el rango completo
        if g.min() != ps or g.max() != n - 1 or len(g) != n - ps:
            raise ValueError(f"{w}: se esperaban las filas {ps} a {n - 1} ({n - ps}); "
                             f"hay {len(g)} entre {g.min()} y {g.max()}")
    return True

def rmse_by_well(pred, truth, wells_df, expected_wells=None):
    """Devuelve (tabla por pozo, RMSE global) sobre el tramo desde el PS.

    truth: columnas well, idx, tvt.
    """
    validate_prediction(pred, wells_df, expected_wells)
    m = pred.merge(truth, on=["well", "idx"], how="left", validate="1:1")
    if m["tvt"].isna().any():
        raise ValueError("hay filas sin TVT real")
    m["se"] = (m["tvt_pred"] - m["tvt"]) ** 2
    por_pozo = m.groupby("well")["se"].agg(n="size", rmse="mean").reset_index()
    por_pozo["rmse"] = np.sqrt(por_pozo["rmse"])
    return por_pozo, float(np.sqrt(m["se"].mean()))

def load_truth(ids_pozos):
    """TVT real desde el PS (solo train)."""
    partes = []
    for w in ids_pozos:
        hz, _, ps = load_well(w)
        partes.append(pd.DataFrame({"well": w, "idx": np.arange(ps, len(hz)),
                                    "tvt": hz["TVT"].to_numpy()[ps:]}))
    return pd.concat(partes, ignore_index=True)

def constant_prediction(tabla):
    """Línea base: TVT constante = último TVT conocido."""
    return pd.concat([pd.DataFrame({"well": r.well, "idx": np.arange(r.ps, r.n),
                                    "tvt_pred": r.tvt_ps})
                      for r in tabla.itertuples()], ignore_index=True)

def to_submission(pred, sample):
    """Convierte al formato de entrega (id, tvt) en el orden de `sample`."""
    s = pd.DataFrame({"id": pred["well"] + "_" + pred["idx"].astype(str),
                      "tvt": pred["tvt_pred"]})
    out = sample[["id"]].merge(s, on="id", how="left", validate="1:1")
    if out["tvt"].isna().any() or len(s) != len(sample):
        raise ValueError("la predicción no coincide con los ids de sample_submission")
    return out

truth = load_truth(wells)
base = constant_prediction(wells_df)
por_pozo, rmse_global = rmse_by_well(base, truth, wells_df, expected_wells=wells)
print("Línea base: TVT constante = TVT(PS)")
print(f"RMSE global (todas las filas): {rmse_global:.2f} ft")
print(f"RMSE por pozo | media: {por_pozo['rmse'].mean():.2f} | mediana: {por_pozo['rmse'].median():.2f}")
```

## 8. Exportar salidas a `data/processed/p1/`

**Exportación**

- Requiere `pyarrow` para escribir parquet (`pip install pyarrow`).
- **Para las partes 2 y 3:** los horizontales de train conservan `TVT` y las columnas de formaciones (`ANCC` … `BUDA`), que no existen en test: no usarlas como entrada del modelo. Tampoco usar `tw_cubre_pozo` ni `tw_cubre_ps`, que se calculan con el TVT real.
- Test se guarda aparte (`horizontal_test/`, `typewell_test/`, `wells_test.parquet`) porque sus ids pueden repetirse en train. Los 3 pozos de la copia local de test también están en train, donde tienen TVT real: posible fuga de información; su resultado no sirve como validación.

```python
def export_split(split, ids_pozos, suf=""):
    (PROCESSED / f"horizontal{suf}").mkdir(parents=True, exist_ok=True)
    (PROCESSED / f"typewell{suf}").mkdir(parents=True, exist_ok=True)
    for w in ids_pozos:
        hz, tw, _ = load_well(w, split)
        clean_gr(hz).to_parquet(PROCESSED / f"horizontal{suf}" / f"{w}.parquet", index=False)
        resample_typewell(tw).to_parquet(PROCESSED / f"typewell{suf}" / f"{w}.parquet", index=False)

export_split("train", wells)
wells_df.to_parquet(PROCESSED / "wells.parquet", index=False)
folds.to_parquet(PROCESSED / "folds.parquet", index=False)

test_wells = sorted(p.name[:8] for p in (RAW / "test").glob("*__horizontal_well.csv"))
export_split("test", test_wells, suf="_test")
wells_test = pd.DataFrame([well_row(w, "test") for w in test_wells])
rad_test = np.radians(wells_test["azimuth"])
wells_test["az_sin"], wells_test["az_cos"] = np.sin(rad_test), np.cos(rad_test)
wells_test.to_parquet(PROCESSED / "wells_test.parquet", index=False)

print("exportados | train:", len(wells), "| test:", len(test_wells),
      "| ids de test que también están en train:", len(set(test_wells) & set(wells)))
```
